```python
#Removes unnecessary spaces ath the beginning and end
df["region"] = df["region"].str.strip().str.title()
```

```python
df["region"] = df["region"].replace({
    "Noth": "North",
    "Northh": "North",
    "Sout": "South",
    "Eastern": "East"
})
```

```python
#if same customer have a known region
customer_region = df.groupby("customer_id")["region"].nunique()

customer_region.value_counts()
```

```python
customer_region[customer_region > 1]
```

```python
#Region mapping create a customer]
customer_region_map = (
    df.dropna(subset=["region"])
      .drop_duplicates("customer_id")
      .set_index("customer_id")["region"]
)
```

```python
#Fill missing regions
df["region"] = df["region"].fillna(
    df["customer_id"].map(customer_region_map)
)
```

```python
#check how many are still missing
df["region"].isnull().sum()
```

```python
df["qty"].describe()
```

```python
#Convert qty and to numeric
df["qty"] = pd.to_numeric(df["qty"], errors="coerce")
df["price"] = pd.to_numeric(df["price"], errors="coerce")
```

```python
#check missing quantity values
df["qty"].isnull().sum()
```

```python
df.loc[df["qty"] < 0, "qty"] = np.nan
df.loc[df["price"] < 0, "price"] = np.nan
```

```python
df["qty"] = df["qty"] * df["price"]
```

```python
#check the missing count
df["qty"].isnull().sum()
```

```python
# View rows where qty is missing
print(df[df["qty"].isnull()])
```

```python
# Percentage of missing qty values
missing_pct = df["qty"].isnull().mean() * 100
print(f"Missing Qty: {missing_pct:.2f}%")
```

```python
#numerical statistics
df["price"].describe()
```

```python
df["price"].isnull().sum()
```

```python
missing_pct = df["price"].isnull().mean() * 100
print(f"Missing Price: {missing_pct:.2f}%")
```

```python
#Clean negative or missing values first
df.loc[df["qty"] < 0, "qty"] = np.nan
df["qty"] = df["qty"].fillna(0)

#calculate revenue
df["total_revenue"] = df["qty"] * df["price"]
```

```python
df["order_date"] = pd.to_datetime(df["order_date"], errors="coerce")
```

```python
import pandas as pd

df = pd.read_csv("data/sales_clean_v1.csv")
```

```python
print(df.head(5))
```

```python
df["total_sales"] = df["qty"] * df["price"]

total_revenue = df["total_sales"].sum()
average_sales = df["total_sales"].mean()
total_quantity = df["qty"].sum()

print(f"Total Revenue: {total_revenue:.2f}$")
print(f"Average Sales Value: {average_sales:,.2f}")
print(f"Total Quantity Sold: {total_quantity:.0f}")
```

```python
#product that generate the most revenue
product_sales = (
    df.groupby("product")["total_sales"]
    .sum()
    .sort_values(ascending=False)
)

product_sales.head(10)
```

```python
#product that have the highest quantity sold
product_quantity = (
    df.groupby("product")["qty"]
    .sum()
    .sort_values(ascending=False)
)

product_quantity.head(10)
```

```python
regional_sales = (
    df.groupby("region")["total_sales"]
    .sum()
    .sort_values(ascending=False)
)

regional_sales
```

```python
regional_quantity = (
    df.groupby("region")["qty"]
    .sum()
    .sort_values(ascending=False)
)

regional_quantity
```

```python
regional_orders = (
    df.groupby("region")["order_id"]
      .nunique()
      .sort_values(ascending=False)
)

regional_orders
```

```python
#Sales Trends Over Time
df["order_date"] = pd.to_datetime(df["order_date"])
```

```python
df["month"] = df["order_date"].dt.to_period("M")
```

```python
#Calculate monthly sales
monthly_sales = (
    df.groupby("month")["total_sales"]
      .sum()
)

monthly_sales
```

```python
#top 10 months/period
monthly_sales.sort_values(ascending=False).head(10)
```

```python
#customer who generated the most revenue
customer_sales = (
    df.groupby("customer_id")["total_sales"]
      .sum()
      .sort_values(ascending=False)
)

customer_sales.head(10)
```

```python
#Customer order frequency
customer_orders = (
    df.groupby("customer_id")["order_id"]
      .nunique()
      .sort_values(ascending=False)
)

customer_orders.head(10)
```

```python
#Highest Value Transactions
df.nlargest(10, "total_sales")[
    ["order_id", "product", "qty", "price", "region", "total_sales"]
]
```

```python
top_transactions = df.nlargest(10, "total_sales")

print(top_transactions[
    ["order_id", "product", "qty", "price", "region", "total_sales"]
])
```

```python
top_transactions = df.nlargest(10, "total_sales")

print(top_transactions[
    ["order_id", "product", "qty", "price", "region", "total_sales"]
])
```

```python
#Monthly Sales Trend
import matplotlib.pyplot as plt
import os

os.makedirs("charts", exist_ok=True)

plt.figure(figsize=(12, 6))

monthly_sales.plot(kind="line", marker="o")

plt.title("Monthly Sales Trend - 2025")
plt.xlabel("Month")
plt.ylabel("Total Sales")
plt.xticks(rotation=45)
plt.grid(True)

plt.tight_layout()
plt.savefig("charts/monthly_sales.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Chart 2: Regional Sales Performance

import matplotlib.pyplot as plt
import os

os.makedirs("charts", exist_ok=True)

plt.figure(figsize=(10, 6))

regional_sales.plot(kind="bar")

plt.title("Regional Sales Performance")
plt.xlabel("Region")
plt.ylabel("Total Sales")
plt.xticks(rotation=0)
plt.grid(axis="y", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig("charts/regional_sales.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Chart 3: Top 10 Products by Revenue

import matplotlib.pyplot as plt
import os

os.makedirs("charts", exist_ok=True)

top_products = product_sales.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))

top_products.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Revenue")
plt.xlabel("Total Sales")
plt.ylabel("Product")
plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig("charts/top_products_revenue.png", dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Chart 4: Top 10 Products by Quantity Sold

import matplotlib.pyplot as plt
import os

os.makedirs("charts", exist_ok=True)

top_quantity = product_quantity.sort_values(ascending=False).head(10)

plt.figure(figsize=(10, 6))

top_quantity.sort_values().plot(kind="barh")

plt.title("Top 10 Products by Quantity Sold")
plt.xlabel("Total Quantity Sold")
plt.ylabel("Product")
plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig("charts/top_products_quantity.png",
            dpi=300, bbox_inches="tight")
plt.show()
```

```python
# Chart 5: Top 10 Customers by Revenue

import matplotlib.pyplot as plt
import os

os.makedirs("charts", exist_ok=True)

top_customers = customer_sales.sort_values(
    ascending=False
).head(10)

plt.figure(figsize=(10, 6))

top_customers.sort_values().plot(kind="barh")

plt.title("Top 10 Customers by Revenue")
plt.xlabel("Total Sales")
plt.ylabel("Customer ID")
plt.grid(axis="x", linestyle="--", alpha=0.5)

plt.tight_layout()
plt.savefig(
    "charts/top_customers_revenue.png",
    dpi=300,
    bbox_inches="tight"
)
plt.show()
```
